```python
print("\033[1m==================================================================\033[0m")
print("\033[1m        SMARTCAMPUS UTILITY & ACCESS PASS GENERATOR\033[0m")
print("\033[1m==================================================================\033[0m")
print()

category=int(input("Select the Category(1:student,2:faculty/staff):"))

if category==1:
    subcategory=(input("Select the Subcategory(UG/PG):"))
    if subcategory=="ug":
        base_fee=500
    elif subcategory=="pg":
        base_fee=350
    else:
        print("[ERROR]:INVALID SUBCATEGORY")
    cgpa=float(input("Enter Student CGPA(0.0-10.0):"))
    if cgpa>=8.5:
        discount=base_fee*0.2
        discount_percent=20
    elif 7.5<=cgpa<=8.49:
        discount=base_fee*0.1
        discount_percent=10
    else:
        discount=0
        discount_percent=0
elif category==2:
    subcategory=(input("Select the Subcategory(Resident/Visiting):"))
    if subcategory=="resident":
        base_fee=800
    elif subcategory=="visiting":
        base_fee=1200
    else:
        print("[ERROR]:INVALID SUBCATEGORY")
    years=int(input("Enter Service Year(0-60):"))
    if years>10:
        base_fee=base_fee*0.15
    else:
        base_fee=base_fee   
else:
    print("[ERROR]:INVALID CATEGORY") 

parking=int(input("Select Parking Permit(0:none,2:two-wheeler,4:four-wheeler):"))
if parking==0:
    parking_fee=0
elif parking==2:
    parking_fee=200
elif parking==4:
    parking_fee=600
else:
    print("[ERROR]:INVALID CHOICE")
if category==1 and parking==4:
    peak_surcharge=150

units=float(input("Enter Monthly Electricity Consumption(in kwh):"))
if units<0:
    print("[ERROR}:Electricity units can not be negative")
elif units<=100:
    electricity_bill=units*3+5
    fixed_charge=50
elif units<=300:
    electricity_bill=(100*3)+((units-100)*5)+100
    fixed_charge=100
elif units<=500:
    electricity_bill=(100*3)+(200*5)+((units-300)*7.5)+150
    fixed_charge=150
else:
    electricity_bill=(100*3)+(200*5)+(200*7.5)+((units-500)*10)+250
    fixed_charge=250
print()
print("-----------------------------------------------------------------")
print("             CALCULATED INVOICE BREAKDOWN")
print("-----------------------------------------------------------------")
print()
print("Base Access Pass Fee :₹",format(base_fee,".2f"))
if discount>0:
    print("Merit Discount("+str(discount_percent)+"%) :-₹",format(discount,".2f"))
else:
    print("Merit Discount :₹0.00")

print("Parking Fee("+str(parking)+") :₹",format(parking_fee,".2f")) 
if peak_surcharge>0:
    print("Student Peak Surcharge :₹",format(peak_surcharge,".2f"))
net_pass_parking=base_fee-discount+parking_fee+peak_surcharge
print("Net Pass & Parking Total :₹",format(net_pass_parking,".2f"))
print("-----------------------------------------------------------------")
print("Electricity Bill("+str(units)+"kWh) :₹",format(electricity_bill,".2f"))
print("Fixed Service Charge :₹",format(fixed_charge,".2f"))
print("-----------------------------------------------------------------")
total_payable=net_pass_parking+electricity_bill
print("TOTAL MONTHLY PAYABLE :₹",format(total_payable,".2f"))
print()
print("\033[1m==================================================================\033[0m")


        
```
